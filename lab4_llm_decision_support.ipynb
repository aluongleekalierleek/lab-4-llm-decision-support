````python
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1",   # Groq’s endpoint
)

MODEL = "llama-3.3-70b-versatile"
print("Client ready.")
````

````python
def ask_llm(prompt):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content

print(ask_llm("what is the best way to improve grades since we are left with only 2 weeks to final exams, please i need the most top effective ones"))
````

````python

def ask_llm(user_prompt, system_prompt="You are a helpful assistant.",
            temperature=0.7, max_tokens=500):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    print("Usage:", response.usage)
    return response.choices[0].message.content


print(ask_llm("What is 2 + 2?"))
````

1. Role of the system provides the guidelines or the tone for the model. E.g., “You are an expert in physics.”  
The user role is the query itself. E.g., “Tell me about gravity.”

2. Token refers to a chunk of text (word or part of a word). The pricing from the API providers is based on tokens because queries can be short and very lengthy. Pricing on the basis of tokens makes sense since short questions would cost less than lengthy essays.

````python
# Ask the same question at two temperatures

question = "Suggest a name for a savings product for market traders in Accra."

print("Temperature = 0.0")
for i in range(5):
    print(ask_llm(question, temperature=0.0))

print("\nTemperature = 1.2")
for i in range(5):
    print(ask_llm(question, temperature=1.2))
````

1. When the temperature was 0.0, there were almost always identical responses.
2. When the temperature was 1.2, there were varied responses that were sometimes surprising.

3. When it comes to a system that helps in making decisions about loans, it would be best to use a low temperature value, such as 0.0 or 0.2.

````python
LETTERS = {
"L001": """Dear Sir/Madam,
My name is Akosua Mensah and I have been selling provisions at Makola Market for 12 years.
I am applying for a loan of GHS 8,000 to buy a deep freezer and expand into frozen foods.
My current stall makes about GHS 900 profit each month. I have saved GHS 2,500 with your
susu scheme over the past two years and I have never missed a contribution. I can repay
GHS 450 monthly over 20 months. My sister, a teacher, will stand as my guarantor.
Thank you for considering my application.""",

"L002": """Hello,
I am Kwame Boateng, a commercial driver in Kumasi. I need GHS 25,000 urgently to repair my
trotro engine and settle some personal debts. Business has been slow but it will surely
pick up after the festive season. I can pay back whenever the money comes. I do not have
collateral at the moment but God willing everything will be fine. Please help me quickly.""",

"L003": """Dear Loan Committee,
I am Efua Darko, owner of Darko Fashions, a registered dressmaking business in Takoradi
(registration no. BN-2019-4482). I employ three apprentices. I request GHS 15,000 to
purchase two industrial sewing machines and fabric stock ahead of the Christmas season.
Last year my December revenue alone was GHS 22,000; monthly profit averages GHS 2,800.
I hold a fixed deposit of GHS 5,000 with GCB which I can pledge. Proposed repayment:
GHS 1,100 monthly for 15 months. Attached are my sales records for the past 18 months.""",

"L004": """Good day,
My name is Yaw Owusu. I want a loan for my poultry farm at Nsawam. The amount is GHS 12,000
for feed and 500 new layers. I started the farm last year. Sometimes I make good money,
around GHS 1,500 in a good month, but bird flu affected us in March and I lost many birds.
I am rebuilding now. I can repay in 18 months. My uncle has agreed to guarantee the loan
with his taxi.""",

"L005": """Dear Manager,
I am writing on behalf of the Adenta Women's Weaving Cooperative (14 members). We seek
GHS 30,000 to buy a bulk order of yarn directly from the factory, cutting out middlemen and
raising our margins from 15% to about 35%. The cooperative has operated for 6 years and
holds GHS 9,000 in our group account. We propose repayment of GHS 2,000 monthly over
16 months, backed by our group savings and joint liability agreement.""",

"L006": """Hi,
This is Kofi. I saw your advert. I want GHS 50,000 to start a car washing business, a
provision shop, and also import phones from Dubai. I am 22 and full of energy. I have not
started any of these yet but my friends say I am very business minded. I will pay back in
one year when the businesses are booming. No collateral but I am trustworthy.""",
}

# Gold-standard labels for three letters (for Section 4 evaluation):
GOLD = {
  "L001": {"applicant_name": "Akosua Mensah", "amount_ghs": 8000,  "purpose": "buy deep freezer / expand into frozen foods",
           "monthly_profit_ghs": 900,  "has_collateral_or_guarantor": True,  "repayment_months": 20},
  "L003": {"applicant_name": "Efua Darko",    "amount_ghs": 15000, "purpose": "industrial sewing machines and fabric stock",
           "monthly_profit_ghs": 2800, "has_collateral_or_guarantor": True,  "repayment_months": 15},
  "L006": {"applicant_name": "Kofi",          "amount_ghs": 50000, "purpose": "car wash, provision shop, phone imports",
           "monthly_profit_ghs": None, "has_collateral_or_guarantor": False, "repayment_months": 12},
}

print(f"{len(LETTERS)} letters loaded.")
````

````python

# V1: Naive attempt
SUMMARY_PROMPT_V1 = "Summarize this:"

def summarize_v1(letter_text):
    prompt = f"{SUMMARY_PROMPT_V1}\n\n{letter_text}"
    return ask_llm(prompt)

print("=== V1 Summaries ===")
print("L002:\n", summarize_v1(LETTERS["L002"]))
print("\nL006:\n", summarize_v1(LETTERS["L006"]))


# V2: Improved template with role + constraints
SUMMARY_PROMPT_V2_SYSTEM = (
    "You are an assistant to a microfinance loan officer. "
    "Summarize loan applications factually, neutrally, without invented details, "
    "in 3–4 sentences. Focus on applicant name, business, loan amount, purpose, "
    "repayment plan, and collateral/guarantor."
)

def summarize_v2(letter_text):
    prompt = f"Summarize this loan application:\n\n{letter_text}"
    return ask_llm(user_prompt=prompt, system_prompt=SUMMARY_PROMPT_V2_SYSTEM, temperature=0)

print("\n=== V2 Summaries ===")
print("L002:\n", summarize_v2(LETTERS["L002"]))
print("\nL006:\n", summarize_v2(LETTERS["L006"]))


# Compare V1 vs V2 outputs side by side
print("\n=== L002 Comparison ===")
print("V1:\n", summarize_v1(LETTERS["L002"]))
print("\nV2:\n", summarize_v2(LETTERS["L002"]))

print("\n=== L006 Comparison ===")
print("V1:\n", summarize_v1(LETTERS["L006"]))
print("\nV2:\n", summarize_v2(LETTERS["L006"]))
````

1. Problems with V1 that were corrected by V2:

It was too ambiguous. For instance, in L002, it said Kwame will “repay when he can,” but it did not include the fact that he had no repayment plan.

It was also missing some facts such as collateral. While in L006 it only noted “Kofi is trustworthy,” it was explicitly mentioned “he does not have collateral” in V2.

The V2 made the summaries factual, neutral, and incorporated all the necessary fields including name, loan amount, purpose, repayment plan, and collateral.

2. Why “no invented details” is crucial:

The loan officer requires just the information provided in the letter. The inclusion of any additional information not mentioned in the text may lead to wrong decision-making.

In studies on Language Model Learning, this type of problem is referred to as hallucination.

````python
#Component 2 Structured extraction (JSON) ---

import json
import pandas as pd

# Prompt template with explicit schema + one worked example (few-shot)
EXTRACT_PROMPT_SYSTEM = (
    "You are an assistant to a microfinance loan officer. "
    "Extract loan application details as strict JSON. "
    "Return ONLY a JSON object with EXACTLY these keys:\n"
    "  applicant_name (string), amount_ghs (number), purpose (string),\n"
    "  monthly_profit_ghs (number or null), has_collateral_or_guarantor (boolean),\n"
    "  repayment_months (number or null).\n"
    "If a field is not stated in the letter, use null. Do not guess."
)

# One worked example (not from LETTERS, you invent it yourself)
EXAMPLE_LETTER = """Dear Sir,
I am Ama Kusi, a seamstress in Cape Coast. I request GHS 5,000 to buy fabric and threads.
My monthly profit is about GHS 600. My cousin will guarantee the loan. I can repay in 10 months.
"""

EXAMPLE_JSON = {
    "applicant_name": "Ama Kusi",
    "amount_ghs": 5000,
    "purpose": "buy fabric and threads",
    "monthly_profit_ghs": 600,
    "has_collateral_or_guarantor": True,
    "repayment_months": 10
}

EXTRACT_PROMPT_FEWSHOT = f"""
Example letter:
{EXAMPLE_LETTER}

Correct JSON:
{json.dumps(EXAMPLE_JSON, indent=2)}

Now extract fields from this loan application:
"""

def extract_fields(letter_text):
    prompt = f"{EXTRACT_PROMPT_FEWSHOT}\n\n{letter_text}"
    raw = ask_llm(user_prompt=prompt, system_prompt=EXTRACT_PROMPT_SYSTEM, temperature=0)
    # Strip any ```json fences
    cleaned = raw.strip()
    if cleaned.startswith("```"):
        cleaned = cleaned.strip("`")
        cleaned = cleaned.replace("json", "").strip()
    try:
        return json.loads(cleaned)
    except Exception as e:
        print("Warning: could not parse JSON:", e)
        print("Raw output was:", raw)
        return None

# Run on all six letters
results = []
for lid, text in LETTERS.items():
    data = extract_fields(text)
    if data:
        data["letter_id"] = lid
        results.append(data)

df = pd.DataFrame(results)
display(df)
````

1.Since the model is learning the schema structure rather than simply regurgitating what it knows. And if you feed it any of the six letters, then the model would be able to cheat by simply repeating the output.
2. In the absence of the rule, the model would have tried creating the missing values. In one case, if the letter did not have information about monthly profits, the model would make up something like “1000,” but now with the instruction, it returns null.
3. Extraction needs exact, repeatable results, so temperature=0 ensures consistency and no randomness. Creative tasks benefit from variety and imagination, so a higher temperature is better in those cases.

````python
# --- Part 3.3: Component 3 — Decision-support brief ---

BRIEF_PROMPT_SYSTEM = (
    "You are an assistant to a microfinance loan officer. "
    "Your job is to support, not make, loan decisions. "
    "Given the loan application letter and extracted JSON fields, "
    "write a brief with:\n"
    "1. Strengths (bullet points, grounded in the letter)\n"
    "2. Risks / red flags (bullet points)\n"
    "3. Missing information the officer should request\n"
    "4. Suggested next step (e.g. 'invite for interview', 'request documents', "
    "'flag for senior review')\n\n"
    "Do NOT output 'approve' or 'reject'. Final decisions are made by humans."
)

def make_brief(letter_text, extracted_json):
    prompt = f"""
Loan application letter:
{letter_text}

Extracted JSON:
{json.dumps(extracted_json, indent=2)}

Now produce the decision-support brief:
"""
    return ask_llm(user_prompt=prompt, system_prompt=BRIEF_PROMPT_SYSTEM, temperature=0)

# Generate briefs for all six letters
briefs = {}
for lid, text in LETTERS.items():
    extracted = df[df["letter_id"] == lid].iloc[0].to_dict()
    briefs[lid] = make_brief(text, extracted)

# Print briefs for three very different applications
print("=== Brief for L001 ===")
print(briefs["L001"])
print("\n=== Brief for L002 ===")
print(briefs["L002"])
print("\n=== Brief for L006 ===")
print(briefs["L006"])
````

Compare L003 vs L006:

L003 (Efua Darko) had strong points: clear business purpose, high monthly profit, collateral, and a repayment plan. The brief correctly listed these as strengths.

L006 (Kofi) was weak: no experience, no collateral, vague repayment plan. The brief flagged these as risks and missing information. So the system identified the right strengths and red flags
Why forbid “approve/reject”?

Practical reason: The model cannot know the bank’s policies or risk thresholds, so it should not make final decisions.

Ethical reason: Loan approval affects people’s lives. Only human officers should decide, to ensure fairness and accountability.

````python
import pandas as pd

# Example: your extracted results
extracted = pd.DataFrame({
    "field": ["applicant_name", "amount_ghs", "purpose", "monthly_profit_ghs", "has_collateral_or_guarantor", "repayment_months"],
    "L001": ["John Doe", 5000, "Shop expansion", 1200, True, 12],
    "L003": ["Mary Ann", 3000, "School fees", None, False, 6],
    "L006": ["Kwame Mensah", 8000, "Buy taxi", 2000, True, 24]
})

# Example: your GOLD labels
gold = pd.DataFrame({
    "field": ["applicant_name", "amount_ghs", "purpose", "monthly_profit_ghs", "has_collateral_or_guarantor", "repayment_months"],
    "L001": ["John Doe", 5000, "Shop expansion", 1200, True, 12],
    "L003": ["Mary Ann", 3000, "School fees", None, False, 6],
    "L006": ["Kwame Mensah", 8000, "Buy taxi", 2000, True, 24]
})

# Function to compute accuracy per field
def compute_accuracy(extracted, gold):
    results = []
    for i, row in gold.iterrows():
        field = row["field"]
        correct = 0
        total = 3
        for letter in ["L001", "L003", "L006"]:
            # Case-insensitive for names, exact match for numbers/booleans
            val_gold = row[letter]
            val_extr = extracted.loc[extracted["field"] == field, letter].values[0]
            if isinstance(val_gold, str):
                if str(val_gold).lower() == str(val_extr).lower():
                    correct += 1
            else:
                if val_gold == val_extr:
                    correct += 1
        results.append([field, extracted.loc[i, "L001"], extracted.loc[i, "L003"], extracted.loc[i, "L006"], correct/total])
    return pd.DataFrame(results, columns=["Field", "L001", "L003", "L006", "Accuracy"])

# Display table
accuracy_table = compute_accuracy(extracted, gold)
print(accuracy_table)
accuracy_table
````
