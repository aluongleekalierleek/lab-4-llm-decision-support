```python
from google.colab import userdata

API_KEY = userdata.get("GROQ_API_KEY")
print("API key loaded successfully!")
```

```python
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1",   # Groq’s endpoint
)

MODEL = "llama-3.3-70b-versatile"
print("Client ready.")
```

```python
def ask_llm(prompt):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content

print(ask_llm("what is the best way to improve grades since we are left with only 2 weeks to final exams, please i need the most top effective ones"))
```

```python

def ask_llm(user_prompt, system_prompt="You are a helpful assistant.",
            temperature=0.7, max_tokens=500):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    print("Usage:", response.usage)
    return response.choices[0].message.content


print(ask_llm("What is 2 + 2?"))
```

1. Role of the system provides the guidelines or the tone for the model. E.g., “You are an expert in physics.”  
The user role is the query itself. E.g., “Tell me about gravity.”

2. Token refers to a chunk of text (word or part of a word). The pricing from the API providers is based on tokens because queries can be short and very lengthy. Pricing on the basis of tokens makes sense since short questions would cost less than lengthy essays.

```python
# Ask the same question at two temperatures

question = "Suggest a name for a savings product for market traders in Accra."

print("Temperature = 0.0")
for i in range(5):
    print(ask_llm(question, temperature=0.0))

print("\nTemperature = 1.2")
for i in range(5):
    print(ask_llm(question, temperature=1.2))
```

1. When the temperature was 0.0, there were almost always identical responses.
2. When the temperature was 1.2, there were varied responses that were sometimes surprising.

3. When it comes to a system that helps in making decisions about loans, it would be best to use a low temperature value, such as 0.0 or 0.2.
